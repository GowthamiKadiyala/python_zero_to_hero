```python
func()
```

```python
def new_decorator(original_func):
    def wrap_func():
        print("Extra code before function")
        original_func()
        print("Extra code after function")
    return wrap_func()
```

```python
def func_needs_decorator():
    print("I want a decorater")
```

```python
func_needs_decorator
```

```python
func_needs_decorator()
```

```python
decorated_func = new_decorator(func_needs_decorator)
```

```python
@new_decorator
def func_needs_decorator():
    print("I want a decorater")
```
