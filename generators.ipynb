```python
def create_cubes(n):
    for x in range(n):
        yield x**3
```

```python
def gen_fibbonacci(n):
    a = 1
    b = 1
    for i in range(n):
        yield a
        a,b = b, a+b
```

```python
for number in gen_fibbonacci(10):
    print(number)
```

```python
def simple_gen():
    for x in range(3):
        yield x
```

```python
for number in simple_gen():
    print(number)
```

```python
import random
random.randint(1,10)
```

```python
def rand_num(low,high,n):
    for i in range(n):
        yield random.randint(low,high)
```

```python
for num in rand_num(1,10,12):
    print(num)
```
